### Activity 1

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import TreebankWordTokenizer

# Only required once
nltk.download('stopwords')

# Use alternative tokenizer
tokenizer = TreebankWordTokenizer()

text = """Natural Language Processing (NLP) is a field of artificial intelligence 
that focuses on the interaction between computers and humans through language."""

# Tokenization
words = tokenizer.tokenize(text)

# Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.lower() not in stop_words and word.isalnum()]

print("Filtered Words:", filtered_words)
```

### Activity 2

```python
# Step 1: Install NLTK (only once)
# !pip install nltk

# Step 2: Import necessary libraries
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import TreebankWordTokenizer

# Step 3: Download required resources
nltk.download('wordnet')
nltk.download('omw-1.4')

# Step 4: Initialize the Lemmatizer and Tokenizer
lemmatizer = WordNetLemmatizer()
tokenizer = TreebankWordTokenizer()

# Step 5: Input text
text = "The striped bats are hanging on their feet for best"

# Step 6: Tokenize the text (no punkt used)
tokens = tokenizer.tokenize(text)

# Step 7: Apply Lemmatization
lemmatized_words = [lemmatizer.lemmatize(word) for word in tokens]

# Step 8: Display Results
print("Original Text:")
print(text)
print("\nLemmatized Words:")
print(lemmatized_words)
```

### Activity 3

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Download required NLTK data (only needs to run once)
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
# Download the missing punkt_tab resource
nltk.download('punkt_tab')


# Initialize lemmatizer
lemmatizer = WordNetLemmatizer()

# Sample text
text = "The striped bats are hanging on their feet for best"

# Tokenize the text
tokens = word_tokenize(text)

# Lemmatize each token
lemmatized_words = [lemmatizer.lemmatize(token) for token in tokens]

# Output result
print("Original Tokens:", tokens)
print("Lemmatized Tokens:", lemmatized_words)
```

### Activity 4

```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk

# Download required NLTK resources
nltk.download('punkt')
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('averaged_perceptron_tagger')
# Download the specific English averaged perceptron tagger resource
nltk.download('averaged_perceptron_tagger_eng')
# Download the required resource for ne_chunk
nltk.download('maxent_ne_chunker_tab')


# Input sentence
sentence = "Barack Obama was born in Hawaii and served as the President of the United States."

# Tokenize and POS tag the sentence
tokens = word_tokenize(sentence)
pos_tags = pos_tag(tokens)

# Perform Named Entity Recognition
named_entities = ne_chunk(pos_tags)

# Display named entities
print(named_entities)
```
